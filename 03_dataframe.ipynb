### 133. DataFrame dropna

### 134. dataframe.dropna (subset)

```python
data = [
    [1, 1, np.nan],
    [10, np.nan, 10],
    [2, 3, 4]
]
columns = ['col0', 'col1', 'col2']
df = pd.DataFrame(data, columns=columns)
df = df.dropna(subset = "col2")
df
```

### 135. dataframe.fillna

### 136. fillna (method)

```python
data = [
    [1, 1, np.nan],
    [10, np.nan, 10],
    [2, 3, 4]
]
columns = ['col0', 'col1', 'col2']
df = pd.DataFrame(data, columns=columns)
df = df.ffill()
df
```
