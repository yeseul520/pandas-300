### 78. 시리즈 슬라이싱

```python
data = [100, 200, 300]
index = ["철수", "영희", "아름"]
s = Series(data, index)
print(s.iloc[:2])
print(s.loc["철수":"영희"])
```

### 79. 시리즈 슬라이싱

```python
data = [100, 200, 300, 400, 500, 600]
index = ["08/01", "08/02", "08/03", "08/04", "08/05", "08/06"]
s = Series(data, index)
print(s)
print(s.index[:3])
print(s.iloc[:3].index)
```

### 80. 시리즈 슬라이싱

```python
data = [100, 200, 300, 400, 500, 600] 
index = ["08/01", "08/02", "08/03", "08/04", "08/05", "08/06"] 
s = Series(data, index)
print(s.iloc[[0,2,4]])
print(s.iloc[0::2])
print(s.loc["08/01":"08/06":2])
```

### 81. 시리즈 추가

### 82. 시리즈 삭제

### 83. 시리즈 삭제 (inplace)
