```python
data = [0, 1, 2, 3, 4, 5]
arr = np.array(data)
arr2 = arr.reshape(2,3)
print(arr2)
print(arr)
```

### 8. reshape

```python
arr = np.array([
    [0, 1, 2],
    [3, 4, 5]
])
arr2 = arr.reshape(6)
print(arr2)
arr2 = arr.flatten()
print(arr2)
arr2 = arr.reshape(-1)
print(arr2)
```
